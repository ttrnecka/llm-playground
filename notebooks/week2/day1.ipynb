# 3 way chat

```python
# imports

import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python

ollama_url = "http://localhost:11434/v1"
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
```

```python
argumentative_system_prompt = """You are Alex, a chatbot who is very argumentative; 
you disagree with anything in the conversation and you challenge everything, in a snarky way. 
You are in a conversation with Blake and Charlie."""

polite_system_prompt = """You are Blake, courteous chatbot. You try to agree with 
everything the other person says, or find common ground. If the other person is argumentative,
you try to calm them down and keep chatting."""

drunk_system_prompt = """You are Charlie, a chatbot who is very drunk. You slur your words and
you don't make much sense, but you are very friendly and keep trying to chat."""

def call_ollama(system_prompt, history):
    messages = [{"role": "system", "content": system_prompt}]
    messages.append({"role": "user", "content": history})
    response = ollama.chat.completions.create(model="llama3.2", messages=messages)
    return response.choices[0].message.content

def build_history(conversation, users):
    user_prompt = f"""
        You are {users[0]}, in conversation with {users[1]} and {users[2]}.
        The conversation so far is as follows:
        {conversation}
        Now with this, respond with what you would like to say next, as Alex.
        """
    return user_prompt
```

```python
full_conversation = "Charlie has spent all his money on drinks and is very drunk."
for i in range(2):
    conversation = call_ollama(polite_system_prompt, build_history(full_conversation, ["Blake", "Alex", "Charlie"]))
    display(Markdown(f"### Blake:\n{conversation}\n"))
    full_conversation += f"Blake: {conversation}\n"

    conversation = call_ollama(argumentative_system_prompt, build_history(full_conversation, ["Alex", "Blake", "Charlie"]))
    display(Markdown(f"### Alex:\n{conversation}\n"))
    full_conversation += f"Alex: {conversation}\n"

    conversation = call_ollama(drunk_system_prompt, build_history(full_conversation, ["Charlie", "Blake", "Alex"]))
    display(Markdown(f"### Charlie:\n{conversation}\n"))
    full_conversation += f"Charlie: {conversation}\n"
```
